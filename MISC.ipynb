```python
def format_indentations(input_path, output_path, start_line, end_line):
    """
    Extracts lines from start_line to end_line (1-indexed, inclusive),
    replaces all tabs with two spaces, and saves the output as a .txt file.
    """
    output_path = 'code_lst/' + output_path + '.txt'
    with open(input_path, 'r', encoding='utf-8') as infile:
        lines = infile.readlines()
    
    # Convert 1-indexed bounds to slice range (inclusive of end_line)
    selected_lines = lines[max(0, start_line - 1):end_line]
    
    # Convert tabs to double spaces
    formatted_lines = [line.replace('    ', '  ') for line in selected_lines]
    
    with open(output_path, 'w', encoding='utf-8') as outfile:
        outfile.writelines(formatted_lines)


format_indentations('train.py', 'train_py', 20, 203)
format_indentations('modules/model_functions.py', 'PerClassMetricsCallback', 673, 764)
format_indentations('modules/data_preprocessing.py', 'calculate_max_diameter', 45, 94)
format_indentations('modules/model_functions.py', 'build_model', 587, 598)
format_indentations('analyze_kits23_tumor_instances.py', 'analyze_kits23_tumor_instances', 4, 24)
format_indentations('apply_tumor_size_limit.py', 'apply_tumor_size_limit', 2, 24)
format_indentations('compile_dataset.py', 'compile_dataset', 4, 66)
format_indentations('modules/model_functions.py', 'get_keras_class_weights', 565, 585)
format_indentations('HPC_out/2026-09-05_06-23/model_evaluation_summary.txt', 'model_evaluation_summary', 1, 15)
```

```python
def plot_example_roc(out_dir):
    fig, ax_roc = plt.subplots(figsize=(9, 8))

    ax_roc.set_title("ROC Curve", fontsize=14, fontweight='bold', pad=10)
    ax_roc.set_xlabel("False Positive Rate")
    ax_roc.set_ylabel("True Positive Rate")
    ax_roc.set_xlim([-0.02, 1.02])
    ax_roc.set_ylim([-0.02, 1.02])

    ax_roc.plot([0, 1], [0, 1], 'k--', lw=2, label="Random Chance (AUC = 0.500)")
    
    ax_roc.plot([0, 0, 1], [0, 1, 1], lw=2, label="Perfect Classifier (AUC = 1.000)", color="#1f77b4")

    ax_roc.legend(loc='lower right', frameon=True)
    ax_roc.grid(True, linestyle='--', alpha=0.6)

    plt.tight_layout()
    save_path = Path(out_dir) / "example_roc_curve.png"
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    plt.close()
    
    print(f"[Success] Example ROC curve saved to: {save_path}")

def plot_example_pr(out_dir):
    fig, ax_pr = plt.subplots(figsize=(9, 8))

    ax_pr.set_title(
        "Precision-Recall Curve", fontsize=14, fontweight="bold", pad=10
    )
    ax_pr.set_xlabel("Recall")
    ax_pr.set_ylabel("Precision")
    ax_pr.set_xlim([-0.02, 1.02])
    ax_pr.set_ylim([-0.02, 1.02])

    ax_pr.plot(
        [0, 1, 1],
        [1, 1, 0],
        drawstyle="steps-post",
        lw=2,
        label="Perfect Classifier (AP = 1.000)",
        color="#1f77b4",
    )

    ax_pr.legend(loc="lower left", frameon=True)
    ax_pr.grid(True, linestyle="--", alpha=0.6)

    plt.tight_layout()
    save_path = Path(out_dir) / "example_pr_curve.png"
    plt.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.close()

    print(f"[Success] Example PR curve saved to: {save_path}")


if __name__ == "__main__":
    output_directory = "./figures"
    os.makedirs(output_directory, exist_ok=True)
    
    plot_example_roc(output_directory)
    plot_example_pr(output_directory)
```
